```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import stata_setup
import os 
STATA_SYSDIR = "/Applications/Stata"
stata_setup.config(STATA_SYSDIR, 'se')
work_dir = "/Users/canyesil/Desktop/simple_macro_AI_replication"
os.chdir(work_dir)
```

```python
# Importing task data

tasks_eloundou = pd.read_csv("data/eloundou_task.csv")[["Task ID",'O*NET-SOC Code','gpt4_automation']]
xwalk_task_occ = pd.read_csv("xwalk/xwalk_eloudou_occ.csv")
easy_hard = pd.read_csv("data/easy_v_hard.csv")
easy_hard["Easy Score"] = easy_hard["Scores"]
easy_hard["Easy Binary"] = 1*(easy_hard["Scores"] > .5)
tasks_eloundou = tasks_eloundou.merge(easy_hard[["Task ID", "Easy Score", "Easy Binary"]], on = "Task ID", how = "left").fillna(0)

# Generate particular metric and aggregate 
tasks_eloundou["automation"] = (tasks_eloundou["gpt4_automation"]>0.5)*tasks_eloundou["gpt4_automation"]
tasks_eloundou["Easy Binary"] = tasks_eloundou["Easy Binary"] * tasks_eloundou["automation"]
tasks_eloundou["Easy Score"] = tasks_eloundou["Easy Score"] * tasks_eloundou["automation"]

# Generating exposure means at 8 digit occupation level
tasks_agg =tasks_eloundou.groupby('O*NET-SOC Code').mean().reset_index(drop=False)
occ_s =[]
for occ in tasks_agg["O*NET-SOC Code"]:
    occ_s.append(occ[0:7])
tasks_agg["OCC_CODE"] = occ_s 
# Generating means at 6 digit occupation levle
tasks_agg = tasks_agg.groupby("OCC_CODE")[["automation","Easy Binary","Easy Score"]].mean().reset_index(drop=False)
tasks_agg
```

```python
# Merging BEA wage data.
j=0
for year in [2022,2021,2020]:
    wages_0 = pd.read_excel("data/wage_data/national_M" + str(year) + "_dl.xlsx")
    wages_0 = wages_0[wages_0["A_MEAN"] != "*"].reset_index(drop=True)
    wages_0 = wages_0[wages_0["TOT_EMP"] != "**"].reset_index(drop=True)
    wages_0["ALL_ECON"]= wages_0["TOT_EMP"][0] * wages_0["A_MEAN"][0]
    col_ = "share_of_emp" + str(year)
    wages_0[col_] = (wages_0["TOT_EMP"] *  wages_0["A_MEAN"]) / wages_0["ALL_ECON"]
    wages_0 = wages_0[wages_0["O_GROUP"]=="detailed"].reset_index(drop=True)
    if j==0:
        wages = wages_0[["OCC_CODE",col_]]
        j = 1
    else:
        wages = wages.merge(wages_0[["OCC_CODE",col_]],on="OCC_CODE",how="outer").fillna(0)

year = 2019
wages_0 = pd.read_excel("data/wage_data/national_M2019_dl.xlsx")
wages_0 = wages_0[wages_0["a_mean"] != "*"].reset_index(drop=True)
wages_0 = wages_0[wages_0["tot_emp"] != "**"].reset_index(drop=True)
wages_0["ALL_ECON"]= wages_0["tot_emp"][0] * wages_0["a_mean"][0]
col_ = "share_of_emp" + str(year)
wages_0[col_] = (wages_0["tot_emp"] *  wages_0["a_mean"]) / wages_0["ALL_ECON"]
wages_0 = wages_0[wages_0["o_group"]=="detailed"].reset_index(drop=True)
wages_0["OCC_CODE"] = wages_0["occ_code"]
wages = wages.merge(wages_0[["OCC_CODE",col_]],on="OCC_CODE",how="outer")
wages =wages.fillna(0)
wages = wages.groupby("OCC_CODE").mean().mean(axis= 1).reset_index(drop=False)
wages["share_of_emp"] = wages[0]
wages = wages[["OCC_CODE","share_of_emp"]]
wages["share_of_emp"] = wages["share_of_emp"]/wages["share_of_emp"].sum()
wages.to_csv("data/merged_wages.csv")
```

```python
# Merging wage data with exposure
vars = ["automation","Easy Binary","Easy Score"]
wages= pd.read_csv("data/merged_wages.csv")
# First get exact matches
wages["OCC_CODE"]=wages["OCC_CODE"].str[:7]
main_6 = wages.groupby("OCC_CODE")["share_of_emp"].sum().reset_index(drop=False).merge(tasks_agg,on="OCC_CODE")
main_6["OCC_CODE"]=main_6["OCC_CODE"].str[:6]
for var in vars:
    main_6[var] = main_6[var] * main_6["share_of_emp"]

main_6 =main_6.groupby("OCC_CODE").sum()
for var in vars:
    main_6[var] = main_6[var] / main_6["share_of_emp"]

print(main_6["share_of_emp"].sum())
print((main_6["automation"] * main_6["share_of_emp"]).sum())

main_6 = main_6.reset_index(drop=False)
tasks_agg["OCC_CODE_5"] = tasks_agg["OCC_CODE"].str[:6]
task_5 =tasks_agg.groupby("OCC_CODE_5")[vars].mean().reset_index(drop=False)
task_append =task_5[-task_5["OCC_CODE_5"].isin(main_6["OCC_CODE"])].reset_index(drop=True)
task_append["OCC_CODE"] = task_append["OCC_CODE_5"]
wages["OCC_CODE"]=wages["OCC_CODE"].str[:6]
exposure = pd.concat([task_append[["OCC_CODE"]+vars],main_6[["OCC_CODE"]+vars]],axis=0).reset_index(drop=True)
exposure.sort_values("OCC_CODE").reset_index(drop=True).to_stata("data/occ5_exp.dta")
```

```python
# Part 1(a): Calculations for Back of the Envelope Numbers
# This section generates the GDP-weighted share of exposed tasks
wages= pd.read_csv("data/merged_wages.csv")
wages["OCC_CODE"]=wages["OCC_CODE"].str[:6]
exposure =pd.read_stata("data/occ5_exp.dta")
main = wages.groupby("OCC_CODE")["share_of_emp"].sum().reset_index(drop=False).merge(exposure, on = "OCC_CODE")

share_exposed =(main["automation"] * main["share_of_emp"]).sum()

print(main["share_of_emp"].sum())
print("GDP-weighted share of tasks exposed to AI:",str((share_exposed * 100)), "percent")
main["Hard Score"] = main["automation"] - main["Easy_Score"]
easy_share_exposed =(main["Easy_Score"] * main["share_of_emp"]).sum()
print("GDP-weighted share of easy for AI tasks:",str((easy_share_exposed * 100)), "percent")
print(easy_share_exposed/share_exposed)
```

```python
# Part 1(b): 
# This section generates labor share values.
# Importing and cleaning data

# Loading data
# First national income
output = pd.read_excel("data/labor_share/bea_nationalincome.xlsx")
integer_list = np.arange(2015,2024)
string_to_add = "nationalincome_"
output.columns = ["index","name"]+ [string_to_add + str(num) for num in integer_list]
# Then wage income
wage = pd.read_excel("data/labor_share/bea_wage.xlsx")
wage = wage.dropna().groupby("bea").sum().reset_index(drop=False)
integer_list = np.arange(2015,2023)
string_to_add = "wage_"
wage.columns = ["index","name"]+ [string_to_add + str(num) for num in integer_list]
# Hours
hours = pd.read_excel("data/labor_share/bea_hours.xlsx")
string_to_add = "hours_"
hours.columns = ["index","name"]+ [string_to_add + str(num) for num in integer_list]
# Number of self-employed by industry
self_emp = pd.read_excel("data/labor_share/bea_selfemployed.xlsx")
string_to_add = "self_emp_"
self_emp.columns = ["index","name"]+ [string_to_add + str(num) for num in integer_list]
# Taxes
taxes = pd.read_excel("data/labor_share/bea_taxes.xlsx")
taxes = taxes.dropna().groupby("index").sum().reset_index(drop=False)
integer_list = np.arange(2015,2023)
string_to_add = "taxes_"
taxes.columns = ["index","name"]+ [string_to_add + str(num) for num in integer_list]
# Hours for self-employed
hours_self = pd.read_stata("data/labor_share/cps_hours.dta")

# We combine everything to obtain labor shares and industry shares (of output)
for year in np.arange(2015,2023):
    year = str(year)
    wage["tot_lab_" + year]= (wage["wage_" + year] + (self_emp["self_emp_" + year] * 1000 * 52 * np.array([40] + list(hours_self["uhrsworkt_w"])) * wage["wage_" + year]/ hours["hours_" + year] / 1000000))/1000
    output["labor_share" + year]= wage["tot_lab_" + year]/ (output["nationalincome_" + year] - (taxes["taxes_" + year]/1000))
    output["ind_share" + year] = output["nationalincome_" + year]/ output.loc[0]["nationalincome_" + year]

string_to_add = "labor_share"
output[string_to_add]=output[[string_to_add + str(num) for num in np.arange(2018,2023)]].mean(axis=1)
string_to_add = "ind_share"
output[string_to_add]=output[[string_to_add + str(num) for num in np.arange(2018,2023)]].mean(axis=1)
print(output.loc[0]["labor_share"])
sL = output[["labor_share","ind_share"]].drop(0).reset_index(drop=True)
sL["NAICS"] = ["11","21","22","23","31-33","42","44-45","48-49","51","52-53","54-56","61-62","71-72","81"]
sL[["NAICS","labor_share","ind_share"]].to_csv("data/labor_shares.csv")
```

```python
# Generating industry-level exposure numbers
xwalk_industry_occ =pd.read_csv("xwalk/xwalk_industry_to_occ.csv").drop("Unnamed: 0", axis = 1)
xwalk_industry_occ = xwalk_industry_occ.T
cols = xwalk_industry_occ.loc["Industry Code"]
xwalk_industry_occ.columns = cols
xwalk_industry_occ = xwalk_industry_occ.drop("Industry Code").drop("Total")
xwalk_industry_occ = xwalk_industry_occ.reset_index(drop=False)
xwalk_industry_occ["OCC_CODE"] = xwalk_industry_occ["index"].str[:6]
of_interest = ["automation"]
transition = xwalk_industry_occ.merge(main[of_interest + ["OCC_CODE"]], on = "OCC_CODE", how ="left")
transition = transition.dropna().reset_index(drop=True)
sums =transition[cols].sum()
for i in range(len(cols)):
    transition[cols[i]] = transition[cols[i]] /sums[cols[i]]

lab_shares = pd.read_csv("data/labor_shares.csv")

for naics in lab_shares["NAICS"]:
    if naics == "31-33":
        transition[naics]= transition["31-330"]
    elif naics == "44-45":
        transition[naics] = transition["44-450"]
    elif naics == "48-49":
        transition[naics] = transition["48-490"]        
    elif len(naics.split("-")) > 1:
        naics_s = []
        for n in (np.arange(int(naics.split("-")[0]),int(naics.split("-")[1])+1,1)*10000):
            naics_s.append(str(n))
        transition[naics] = transition[naics_s].sum(axis=1)
    else:
        transition[naics]= transition[str(int(naics)*10000)]

for i in ["52-53","54-56","61-62","71-72"] :
    transition[i] = transition[i] /(transition[i].sum())

rel_cols = ["OCC_CODE","automation"] + list(lab_shares["NAICS"])
transition_final =transition[rel_cols]

industries = []
cols =list(lab_shares["NAICS"])
for col in cols:
    exp = []
    for index in of_interest:
        exp.append((transition_final[index] * transition_final[col]).sum())

    industries.append(exp)

exposure_industry =pd.DataFrame(cols)
exposure_industry= pd.concat([exposure_industry, pd.DataFrame(industries)], axis = 1)
exposure_industry.columns = ["NAICS","automation"] 
exposure_industry

exposure_industry["labor_share"] = lab_shares["labor_share"]
exposure_industry["ind_share"] = lab_shares["ind_share"]
all = exposure_industry

for index in ["automation"]:
    all["contribution"]=(all[index] * all["ind_share"])
    all["share_weight"]=all["contribution"]/ all["contribution"].sum()
    right_labor_share = (all["share_weight"] * all["labor_share"]).sum()
    print("Labor Share of Income for Exposed Tasks", right_labor_share)
```

```python
print("Double Checking Baseline Case Results:")
print()
print("GDP Weighted Share of Exposed Tasks",share_exposed)
print("Labor Share of Income for Exposed Tasks",right_labor_share)
print("Productivity:",right_labor_share * .27)
print("Exposed Feasible Task Share:",share_exposed * .23)
print("TFP:",share_exposed * .23 * right_labor_share * .27)

print("Easy v Hard Results:")
print("GDP Weighted Share of Easy Exposed Tasks",easy_share_exposed)
print("Share of Easy Tasks Among Exposed:", str(100* easy_share_exposed/share_exposed)+"%") 
print("TFP:",(easy_share_exposed * .27 + (share_exposed-easy_share_exposed) * .07)*.23 * right_labor_share)
```
